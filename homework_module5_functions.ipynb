## Перевірка на просте число

- Приймає число

Просте число ділиться тільки на 1 та на себе.

Перевірте числа: 2, 3, 4, 17, 20, 29

````python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
            
    return True
print(is_prime(2))
print(is_prime(3))
print(is_prime(4))
print(is_prime(17))
print(is_prime(20))
print(is_prime(29))
````

- Приймає число n

## Комбінація параметрів
Створіть функцію `calculate(operation, *numbers, **options)` яка:
- operation: "sum", "multiply", "max", "min"
- numbers: будь-яка кількість чисел
- options: додаткові параметри (наприклад, round=True)

Виконайте операцію над числами.

Приклади:
- calculate("sum", 1, 2, 3, 4, 5)
- calculate("multiply", 2, 3, 4)
- calculate("max", 10, 25, 7, 30, 15)

````python
def calculate(operation, *numbers, **options):
    if operation == "sum":
        return sum(numbers)
    elif operation == "multiply":
        res = 1 
        for i in numbers:
            res *= i
    elif operation == "max":
        res =  max(numbers)
    elif operation == "min":
        res =  min(numbers)
    else:
        return "невідома операція"
    
    if options.get("round") == True:
        res = round(res)
    return res
print(calculate("sum", 1, 2, 3, 4, 5))
print(calculate("multiply", 2, 3, 4))
print(calculate("max", 10, 25, 7, 30, 15))
````

## Факторіал рекурсивно
Створіть рекурсивну функцію `factorial_recursive(n)` яка:
- Базовий випадок: якщо n <= 1, повернути 1
- Рекурсивний: n × factorial(n-1)

Обчисліть: 5!, 7!

````python
def factorial_recursive(n):
    if n <= 1:
        return 1 
    return n * factorial_recursive(n - 1)
print(factorial_recursive(5))
````

## Фібоначчі рекурсивно
Створіть рекурсивну функцію `fibonacci(n)` яка:
- Базові випадки: fib(0)=0, fib(1)=1
- Рекурсивний: fib(n-1) + fib(n-2)

Виведіть перші 10 чисел Фібоначчі.

````python
def fibonacci(n):
    if n == 0 :
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2) 
for i in range(10):
    print(fibonacci(i))
````

## Сума цифр рекурсивно
Створіть рекурсивну функцію `sum_digits(n)` яка:
- Обчислює суму цифр числа
- Базовий: якщо n < 10, повернути n
- Рекурсивний: (n % 10) + sum_digits(n // 10)

Протестуйте: 123 → 6, 9875 → 29

````python
def sum_digits(n):
    if n < 10:
        return n
    return sum_digits(n % 10) + sum_digits(n // 10)
print(sum_digits(123))
print(sum_digits(9875))
````

## Степінь рекурсивно
Створіть рекурсивну функцію `power(base, exp)` яка:
- Обчислює base^exp
- Базовий: якщо exp == 0, повернути 1
- Рекурсивний: base × power(base, exp-1)

Обчисліть: 2^5, 3^4

````python
def power(base, exp):
    if exp == 0:
        return 1 
    return base * power(base, exp - 1) 
print(power(2, 5))
print(power(3, 4))
````

## НСД рекурсивно (алгоритм Евкліда)
Створіть рекурсивну функцію `gcd(a, b)` яка:
- Знаходить найбільший спільний дільник
- Базовий: якщо b == 0, повернути a
- Рекурсивний: gcd(b, a % b)

Знайдіть НСД: (48, 18), (100, 35)

````python
def gcd(a, b):
    if b == 0:
        return a 
    return gcd(b, a % b)
print(gcd(48, 18))
print(gcd(100, 35))
````

Поверніть True/False.

Протестуйте:
- "test@example.com" → True
- "invalid" → False
- "test @mail.com" → False

Створіть функцію `generate_password(length=8, include_digits=True, include_special=False)` яка:
- Генерує випадковий пароль заданої довжини
- include_digits: додає цифри 0-9
- include_special: додає спецсимволи !@#$%

Підказка: використайте модуль `random` та `string`

````python
import random as r
import string as s
def generate_password(length=8, include_digits=True, include_special=False):
    chars = s.ascii_letters
    
    if include_digits:
        chars += s.digits
    if include_special:
        chars += "!@#$%"
        
    return ''.join(r.choice(chars) for _ in range(length))
print(generate_password())
````

## Групування даних
Створіть функцію `group_by(items, key_func)` яка:
- Приймає список та функцію для отримання ключа
- Повертає словник {ключ: [елементи]}

Приклад:
```python
words = ["apple", "ant", "ball", "bat", "cat"]
grouped = group_by(words, lambda x: x[0])
# {'a': ['apple', 'ant'], 'b': ['ball', 'bat'], 'c': ['cat']}
```

````python
def group_by(items, key_func):
    res = {}
    for i in items:
        key = key_func(i)

        if key not in res:
            res[key] = []

        res[key].append(i)

    return res
words = ["apple", "ant", "ball", "bat", "cat"]

grouped = group_by(words, lambda x: x[0])

print(grouped)
````

data = [-5, 3, -2, 7, 1, -8, 4]
